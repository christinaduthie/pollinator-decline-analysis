```python
# List of CSV files to be combined
csv_files = [
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2014.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2015.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2016.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2017.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2018.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2019.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2020.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2021.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2022.csv',
    '/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/monarch_sightings_2023.csv'
]
```

```python
# List to hold dataframes
dataframes = []

# Loop through each CSV file
for file in csv_files:
    # Read the CSV file into a DataFrame
    df = pd.read_csv(file)
    
    # Try to convert the 'Date' column to datetime with a specific format, adjust as per the format in your files
    # Example format: 'YYYY-MM-DD'
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
    # Append the cleaned dataframe to the list
    dataframes.append(df)

# Concatenate all the dataframes into one
combined_df = pd.concat(dataframes, ignore_index=True)

# Save the combined dataframe to a new CSV file
combined_df.to_csv('/Users/duthie/Desktop/MyProjects/DataScience Projects/pollinator-decline-analysis/Data/butterfly_dataset/combined_monarch_sightings_2014_2023.csv', index=False)

print("Data combined and saved successfully to 'combined_monarch_sightings_2014_2023.csv'.")
```
